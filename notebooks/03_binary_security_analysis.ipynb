## 1 — Binary experiment via the packaged runner

```python
from aghc_s128.experiments import load_config, run_binary_experiments
config = load_config('configs/binary_experiments.yaml')
config['password'] = 'aghc-s128-demo'  # public test vector
payload = run_binary_experiments(config)
print('round trips ok:', sum(r['round_trip_ok'] for r in payload['runs']),
      '/', len(payload['runs']))
print('outputs:', payload['outputs'])
```

## 2 — Metrics table

```python
rows = [{'input': r['input'], **r['metrics']} for r in payload['runs']]
try:
    import pandas as pd  # optional dependency, pretty display only
    display(pd.DataFrame(rows))
except ImportError:
    for row in rows:
        print(row)
```

## 3 — Integrity check (SHA3-256 round trip)

```python
for run in payload['runs']:
    print(run['input'], '->', 'OK' if run['round_trip_ok'] else 'FAILED')
```
