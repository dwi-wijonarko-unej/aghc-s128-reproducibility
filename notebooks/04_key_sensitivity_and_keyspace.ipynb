## 1 — Mode B: key sensitivity (two passwords, one plaintext)

```python
from aghc_s128.experiments import load_config, run_key_sensitivity
config = load_config('configs/key_sensitivity.yaml')
config['password_a'] = 'aghc-s128-demo-key-A'  # public test vectors
config['password_b'] = 'aghc-s128-demo-key-B'
payload = run_key_sensitivity(config)
for key, value in payload['metrics'].items():
    print(f'{key:40s} {value:.4f}')
print('figure:', payload['outputs']['figure'])
```

Interpretation note: high key sensitivity / NPCR / bit difference indicate the two ciphertexts differ; they are statistical observations, not a security proof.

## 2 — Mode C: password key-space estimation

```python
from aghc_s128.experiments import keyspace_analysis
result = keyspace_analysis(password_length=16, charset_size=95)
print('keyspace     :', result['keyspace_scientific'])
print('entropy bits :', round(result['password_entropy_bits'], 2))
print('classification:', result['password_entropy_classification'])
print()
print(result['caveat'])
```
